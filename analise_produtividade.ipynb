# Análise Exploratória de Dados de Produtividade

Este projeto apresenta uma análise exploratória de um conjunto de dados sobre características de funcionários e produtividade. O trabalho utiliza Python para organizar, explorar e visualizar os dados, com foco em variáveis demográficas, profissionais e indicadores de produtividade.

## Objetivos

- identificar e classificar as variáveis do conjunto de dados;
- explorar a distribuição das variáveis qualitativas;
- criar visualizações para facilitar a interpretação dos dados;
- investigar a distribuição de produtividade entre diferentes grupos.

## Ferramentas

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn

## Dados

O conjunto de dados possui 230 observações e informações como idade, gênero, departamento, salário, horas trabalhadas, produtividade, satisfação, tempo na empresa, cursos realizados e modalidade de trabalho. Uma variável binária de produtividade é construída a partir da mediana do indicador de produtividade.


## Carregamento dos dados


```python
import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/cibelerusso/Datasets/main/DadoseDecisoes.csv'

df = pd.read_csv(url)
```

```python
mediana = df['Produtividade'].median()

#se a produtividade for maior que a mediana ele receberá 1, caso contrário
#receberá 0
df['Produtividade_bin'] = np.where(df['Produtividade']>mediana, 1, 0)

df
```

### Classificação das variáveis


ID - Quantitativa discreta

Idade - Quantitativa discreta

Gênero - Qualitativa nominal

Departamento - Qualitativa nominal

Salário - Quantitativa contínua

Horas_Trabalhadas - Quantitativa discreta

Produtividade - Quantitativa contínua

Satisfação - Quantitativa contínua

Tempo_Empresa - Quantitativa discreta

Cursos_realizados - Quantitativa discreta

Home_Office - Qualitativa nominal

Produtividade_bin - Qualitativa nominal


### Distribuição das variáveis qualitativas

Vamos gerar tabelas de frequência para as variáveis qualitativas.


```python
# Tabela de Frequência para Gênero
print("Tabela de Frequência para Gênero:")
display(df['Gênero'].value_counts().to_frame())

# Tabela de Frequência para Departamento
print("\nTabela de Frequência para Departamento:")
display(df['Departamento'].value_counts().to_frame())

# Tabela de Frequência para Home_Offic
print("\nTabela de Frequência para Home_Office:")
display(df['Home_Office'].value_counts().to_frame())

# Tabela de Frequência para Produtividade_bin
print("\nTabela de Frequência para Produtividade_bin:")
display(df['Produtividade_bin'].value_counts().to_frame())
```

### Visualização das variáveis qualitativas

Agora, vamos visualizar a distribuição das variáveis qualitativas usando gráficos de barras.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Definir um estilo para os gráficos
sns.set_style('whitegrid')

# Criar subplots para cada variável qualitativa
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
axes = axes.flatten()

vars_quali = ['Gênero', 'Departamento', 'Home_Office', 'Produtividade_bin']

for i, col in enumerate(vars_quali):
    sns.countplot(x=col, data=df, ax=axes[i], palette='viridis', hue=col, legend=False)
    axes[i].set_title(f'Distribuição de {col}', fontsize=14)
    axes[i].set_xlabel(col, fontsize=12)
    axes[i].set_ylabel('Contagem', fontsize=12)
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

### Descrição para as variáveis quantitativas



```python
vars_quant = ['Idade', 'Salário','Horas_Trabalhadas' , 'Produtividade', 'Satisfação', 'Tempo_Empresa', 'Cursos_Realizados']

df[vars_quant].describe()
```

É interessante observar os quartis da vriável Salário. Vemos que muitos funcionários ganham pouco enquanto poucos ganham muito. Esse fato será melhor analisado mais tarde.

### Box Plots para Variáveis Quantitativas

Vamos visualizar a distribuição das variáveis quantitativas usando box plots, que nos ajudarão a identificar a dispersão, a mediana e possíveis outliers.

```python

sns.set_style('whitegrid')

# Criar subplots para cada variável quantitativa
num_vars_quant = len(vars_quant)
fig, axes = plt.subplots(nrows=(num_vars_quant + 1) // 2, ncols=2, figsize=(15, 5 * ((num_vars_quant + 1) // 2)))
axes = axes.flatten()

for i, col in enumerate(vars_quant):
    # Use 'color' instead of 'palette' for single box plots to avoid FutureWarning
    sns.boxplot(y=df[col], ax=axes[i], color=sns.color_palette('viridis')[0])
    axes[i].set_title(f'Box Plot de {col}', fontsize=14)
    axes[i].set_ylabel(col, fontsize=12)

# Remover subplots vazios, se houver
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

### Interpretação dos Box Plots


Vamos analisar cada variável:

*   **Idade:** A distribuição da idade parece razoavelmente simétrica, com a maioria dos funcionários concentrada entre 20 e 40 anos. Há alguns outliers na faixa etária mais alta, o que é comum.

*   **Salário:** O box plot de 'Salário' mostra uma distribuição bastante assimétrica e com muitos outliers na parte superior. Isso sugere que a maioria dos funcionários tem salários mais baixos, enquanto um pequeno número de funcionários tem salários significativamente mais altos. A mediana é relativamente baixa em comparação com o valor máximo.

*   **Horas_Trabalhadas:** A distribuição das horas trabalhadas por semana parece estar bem concentrada, com a maioria dos funcionários trabalhando entre 30 e 45 horas. Existem alguns outliers indicando indivíduos que trabalham horas incomuns.

*   **Produtividade:** A 'Produtividade' parece ter uma distribuição mais ou menos simétrica, mas com alguns valores atípicos tanto na parte inferior quanto na superior, indicando funcionários com produtividade muito alta ou muito baixa em comparação com a maioria.

*   **Satisfação:** A variável 'Satisfação' mostra uma distribuição bem concentrada na parte superior da escala, indicando que a maioria dos funcionários está satisfeita. Há alguns outliers na parte inferior, representando funcionários com níveis de satisfação significativamente baixos.

*   **Tempo_Empresa:** 'Tempo_Empresa' apresenta uma distribuição assimétrica, com a maioria dos funcionários tendo menos tempo de casa. Há outliers indicando funcionários com um tempo de empresa consideravelmente longo.

*   **Cursos_Realizados:** A distribuição de 'Cursos_Realizados' também é assimétrica, com a maioria dos funcionários tendo feito poucos cursos. Existem outliers de funcionários que realizaram um número elevado de cursos.

### Histogramas para Variáveis Quantitativas

Vamos agora visualizar a distribuição de cada variável quantitativa utilizando histogramas. Os histogramas nos ajudam a entender a frequência de cada valor ou faixa de valores em uma variável, revelando padrões como assimetria, curtose e a presença de múltiplos picos (multimodalidade).

```python

sns.set_style('whitegrid')

# Criar subplots para cada variável quantitativa
num_vars_quant = len(vars_quant)
fig, axes = plt.subplots(nrows=(num_vars_quant + 1) // 2, ncols=2, figsize=(15, 5 * ((num_vars_quant + 1) // 2)))
axes = axes.flatten()

for i, col in enumerate(vars_quant):
    sns.histplot(df[col], kde=True, ax=axes[i], color='rebeccapurple')
    axes[i].set_title(f'Histograma de {col}', fontsize=14)
    axes[i].set_xlabel(col, fontsize=12)
    axes[i].set_ylabel('Frequência', fontsize=12)

# Remover subplots vazios, se houver
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

### Análise de Associações com `Produtividade_bin`

Agora vamos investigar as possíveis associações entre a variável `Produtividade_bin` e as outras variáveis do dataset.

#### Associações com Variáveis Qualitativas

Vamos começar com as variáveis qualitativas, criando tabelas de contingência e visualizando as proporções de `Produtividade_bin` para cada categoria.

```python
# Variáveis qualitativas (excluindo 'Produtividade_bin' e 'ID')
quali_vars_for_assoc = ['Gênero', 'Departamento', 'Home_Office']

for col in quali_vars_for_assoc:
    print(f"\nTabela de Contingência para {col} vs Produtividade_bin:")
    contingency_table = pd.crosstab(df[col], df['Produtividade_bin'], margins=True)
    display(contingency_table)

    print(f"Proporções de Produtividade_bin por {col}:")
    proportions = pd.crosstab(df[col], df['Produtividade_bin'], normalize='index') * 100
    display(proportions)

    # Plotar gráficos de barras agrupados
    plt.figure(figsize=(8, 5))
    sns.countplot(x=col, hue='Produtividade_bin', data=df, palette='viridis')
    plt.title(f'Distribuição de Produtividade_bin por {col}', fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel('Contagem', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='Produtividade_bin', labels=['Baixa', 'Alta'])
    plt.tight_layout()
    plt.show()
```

#### Associações com Variáveis Quantitativas

Agora, vamos analisar a relação entre `Produtividade_bin` e as variáveis quantitativas usando box plots, que nos permitirão comparar a distribuição de cada variável quantitativa entre os dois grupos de produtividade (baixa e alta).

```python
sns.set_style('whitegrid')


num_vars_quant = len(vars_quant)
fig, axes = plt.subplots(nrows=(num_vars_quant + 1) // 2, ncols=2, figsize=(15, 5 * ((num_vars_quant + 1) // 2)))
axes = axes.flatten()


for i, col in enumerate(vars_quant):
    sns.boxplot(x='Produtividade_bin', y=col, data=df, ax=axes[i], palette='viridis', hue='Produtividade_bin', legend=False)
    axes[i].set_title(f'{col} por Produtividade_bin', fontsize=14)
    axes[i].set_xlabel('Produtividade_bin (0=Baixa, 1=Alta)', fontsize=12)
    axes[i].set_ylabel(col, fontsize=12)

# Remover subplots vazios, se houver
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.legend(title='Produtividade_bin', labels=['Baixa', 'Alta'])

plt.show()
```

### Analisando os gráficos

* Em geral, os departamentos tem Produtividade_bin baixa. Em especial, vemos que o 60% do departamento de RH tem produtividade abaixo da mediana.

* Vemos também que, todo funcionário que **não** trabalha em Home Office apresenta produtividade_bin baixa, enquanto 99% dos funcionários que trabalham em Home Office apresentam produtividade alta.

* Funcionários com produtividade baixa tendem a ter menos cursos realizados, com a mediana de cursos estando em um valor inferior. Por outro lado, funcionários com produtividade alta geralmente realizaram mais cursos, com a mediana em um valor superiror. Isso **sugere** uma relação positiva entre quantidade de cursos realizados e produtividade.

* Funcionários com mais tempo de empresa apresentam uma mediana similar a dos funcionários com menos tempo de empresa. Isso **sugere** que o tempo de empresa não é um fator que diferencia significativamente a produtividade dos funcionários.

* Sobre a satisfação dos funcionários e suas produtividade, vemos que a mediana dos dois grupos são similares.

* Observando o boxplot Salário por Produtividade_bin parece haver uma grande sobreposição, isso sugere que ganhar mais não é sempre indicativo de ser mais produtivo.

* Os funcionários de baixa produtividade tendem a trabalhar menos horas, com a mediana bem menor que a mediana dos funcionários de alta produtividade. Os funcionários de alta produtividade apresentam um primeiro quartil maior comparado com os funcionários de baixa produtividade. Isso indica que, mesmo entre os 25% com menos horas trabalhadas em cada grupo, os de alta produtividade ainda trabalham mais horas.




### Variáveis com Associação Mais Forte com Produtividade_bin

Com base nas análises que fizemos e os gráficos plotados, as variáveis que parecem ter uma associação mais forte com `Produtividade_bin`:

1.  **Home_Office:** Esta é a variável com a associação mais notável. A análise revelou que **todos os funcionários que não trabalham em Home Office apresentam produtividade_bin baixa**, enquanto aproximadamente **99% dos funcionários que trabalham em Home Office apresentam produtividade alta**.

2.  **Horas_Trabalhadas:** Existe uma clara distinção entre os grupos de produtividade. Funcionários com baixa produtividade tendem a trabalhar menos horas, com uma mediana significativamente menor do que a dos funcionários de alta produtividade. Além disso, o primeiro quartil para os funcionários de alta produtividade é maior do que o dos de baixa produtividade, indicando que mesmo os 25% menos produtivos no grupo de alta produtividade ainda trabalham mais horas que o grupo de baixa produtividade.

3.  **Cursos_Realizados:** Há uma relação positiva sugerida. Funcionários com produtividade baixa geralmente realizaram menos cursos (com uma mediana inferior), enquanto aqueles com produtividade alta realizaram mais cursos (com uma mediana superior). Isso indica que a realização de cursos **pode** estar ligada a uma maior produtividade.

4.  **Departamento:** Embora não seja tão categórica quanto Home_Office, o `Departamento` também mostra uma associação, especialmente no departamento de RH, onde aproximadamente **60% dos funcionários apresentam produtividade abaixo da mediana**. Isso sugere que o departamento pode influenciar a produtividade.

As variáveis `Tempo_Empresa` e `Satisfação` não apresentaram diferenças significativas nas medianas entre os grupos de produtividade baixa e alta, **sugerindo** uma associação mais fraca ou inexistente com a `Produtividade_bin`.




### Limitações da análise ###
